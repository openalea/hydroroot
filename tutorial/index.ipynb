# 🌿 Modeling Plant Hydraulic Architecture

## 📌 Scientific Context

This project aim to model root hydraulic architecture in OpenAlea.
The aim is to simulate hydraulic fluxes in an explicit root system architecture (RSA), represented as a Multiscale Tree Graph (MTG).

OpenAlea.HydroRoot allows to  :
- Recontruct or simulate root architecture as a MTG 
- Set or infer the distribution of axial and radial hydraulic properties
- Simulate water fluxes inside the RSA using hydrostatic or solute solvers
- Analyse and Visualize the role of architecture and hydraulic properties on transport
---

## 🧠 Methodological Workflow

OpenAlea.HydroRoot rely on :

- RSA representation as a MTG (Multi-scale Tree Graph)
- Reading RSA from RSML format or simulating RSA using Markov Models
- Assignement of hydraulic properties to the plant segments
- Solving flow equations using electrical analogies (Kirchhoff's laws) or using solute transport equations using continuous medium analogy
- Steady-state and dynamic simulations
